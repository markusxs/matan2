$$ 
f(x) =  
\begin{cases}
x / 2, & x \in [0, 2); \\
1, & x \in [2, 4].
\end{cases}
$$

```python
from numpy import cos, sin, pi, linspace, where
import matplotlib.pyplot as plt
```

```python
def S1_n(n, x):
    S = 3/4
    k = 1
    while k <= n:
        S += ((-1)**k - 1) * cos(k * pi * x / 2) / (k**2 * pi**2)
        S -= k * pi * x / (2 * k * pi)
        k += 1
    return S
```

```python
def S2_n(n, x):
    S = 3/4
    k = 1
    while k <= n:
        S += 4 * (cos(k * pi / 2) - 1) * cos(k * pi * x / 4) / (k**2 * pi **2)
        k += 1
    return S
```

```python
def S3_n(n, x):
    S = 0
    k = 1
    while k <= n:
        m = 4 * sin(k * pi / 2) / (k**2 * pi**2) - 2 * (-1)**k / (k * pi)
        S += m * sin(k * pi * x / 4)
        k += 1
    return S
```

```python
def setAndCustomise(g, X, Y):
    g.plot(X,Y, color="b")
    g.grid(True, linestyle='--', alpha=0.5)
    g.axhline(0, color="black", alpha=0.5)
    g.axvline(0, color="black", alpha=0.5)
```

```python
def f(x):
    if 0 <= x < 2: return x / 2
    if 2 <= x <= 4: return 1

X = linspace(0, 4, 100)
Y = where(X <= 2, X/2, 1)
setAndCustomise(plt, X, Y)
```

```python
N = [2, 10, 30]
Ses = [["1. Общий тригонаметрический ряд Фурье", S1_n], ["2. Ряд Фурье по косинусам", S2_n], ["3. Ряд Фурье по синусам", S3_n]]

X = linspace(-5, 10, 200)
for a, S in Ses:
    fig, axes = plt.subplots(1, 3, figsize=(21, 7))
    print(a + " для n:", *N)
    for i in range(3):
        n = N[i]
        ax = axes[i]
        Y = S(n, X)
        ax.plot(X, Y)
    
        setAndCustomise(ax, X, Y)
        ax.set_xlabel('X', fontsize=10)
        ax.set_ylabel('Y', fontsize=10)
```
